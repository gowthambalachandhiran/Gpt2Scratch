## We will now do simple tokenization

# - First we will split the sentence to words
# - Now we then get unique set of words from our corpous
# - We assign an integer to the the words
# - We will write a method which will given a piece of text will encode
# - We will also write a method to decode the encoded text

image.png


```python
#Splitting the raw text as simple tokens
preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', raw_text)
preprocessed = [item.strip() for item in preprocessed if item.strip()]
print(preprocessed[:30])
print(len(preprocessed))
#Sort it and only keep the unique words
all_words = sorted(set(preprocessed))
vocab_size = len(all_words)
#Get vacob size
print(vocab_size)
#Map integer to words
vocab = {token:integer for integer,token in enumerate(all_words)}
```

```python
class SimpleTokenizerV1:
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = {i:s for s,i in vocab.items()}

    def encode(self, text):
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)

        preprocessed = [
            item.strip() for item in preprocessed if item.strip()
        ]
        ids = [self.str_to_int[s] for s in preprocessed]
        return ids

    def decode(self, ids):
        text = " ".join([self.int_to_str[i] for i in ids])
        # Replace spaces before the specified punctuations
        text = re.sub(r'\s+([,.?!"()\'])', r'\1', text)
        return text
```

```python
tokenizer = SimpleTokenizerV1(vocab)

text = input()
ids = tokenizer.encode(text)
print(ids)
```

```python
tokenizer.decode(ids)
```

## There are special tokens to handle OOV or unknown tokens from vocabolory

Some of these special tokens are

[BOS] (beginning of sequence) marks the beginning of text
[EOS] (end of sequence) marks where the text ends (this is usually used to concatenate multiple unrelated texts, e.g., two different Wikipedia articles or two different books, and so on)
[PAD] (padding) if we train LLMs with a batch size greater than 1 (we may include multiple texts with different lengths; with the padding token we pad the shorter texts to the longest length so that all texts have an equal length)
[UNK] to represent words that are not included in the vocabulary

Note that GPT-2 does not need any of these tokens mentioned above but only uses an <|endoftext|> token to reduce complexity

The <|endoftext|> is analogous to the [EOS] token mentioned above

GPT also uses the <|endoftext|> for padding (since we typically use a mask when training on batched inputs, we would not attend padded tokens anyways, so it does not matter what these tokens are)

GPT-2 does not use an <UNK> token for out-of-vocabulary words; instead, GPT-2 uses a byte-pair encoding (BPE) tokenizer, which breaks down words into subword units which we will discuss in a later section

We use the <|endoftext|> tokens between two independent sources of text:

## We will implement the special token for unknown vocab



```python
class SimpleTokenizerV2:
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = { i:s for s,i in vocab.items()}

    def encode(self, text):
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
        preprocessed = [item.strip() for item in preprocessed if item.strip()]
        preprocessed = [
            item if item in self.str_to_int
            else "<|unk|>" for item in preprocessed
        ]

        ids = [self.str_to_int[s] for s in preprocessed]
        return ids

    def decode(self, ids):
        text = " ".join([self.int_to_str[i] for i in ids])
        # Replace spaces before the specified punctuations
        text = re.sub(r'\s+([,.:;?!"()\'])', r'\1', text)
        return text
```

```python
tokenizer_2 = SimpleTokenizerV2(vocab)
text = input()
ids = tokenizer_2.encode(text)
print(ids)
tokenizer_1 = SimpleTokenizerV1(vocab)
ids = tokenizer_1.encode(text)
print(ids)
```

## Now we would see what is byte pair encoding

BytePair encoding
GPT-2 used BytePair encoding (BPE) as its tokenizer
it allows the model to break down words that aren't in its predefined vocabulary into smaller subword units or even individual characters, enabling it to handle out-of-vocabulary words
For instance, if GPT-2's vocabulary doesn't have the word "unfamiliarword," it might tokenize it as ["unfam", "iliar", "word"] or some other subword breakdown, depending on its trained BPE merges
The original BPE tokenizer can be found here: https://github.com/openai/gpt-2/blob/master/src/encoder.py
In this chapter, we are using the BPE tokenizer from OpenAI's open-source tiktoken library, which implements its core algorithms in Rust to improve computational performance
I created a notebook in the ./bytepair_encoder that compares these two implementations side-by-side (tiktoken was about 5x faster on the sample text)

## I will walk you through the Byte Pair Encoding from scratch

# BPE Algorithm

#- To explain the Byte Pair Encoding (BPE) algorithm, let's use a new, slightly more complex sentence that demonstrates how common sub-words are compressed:

# -Sentence: banana bandanaIn BPE, we start by treating every single character (including spaces) as a unique token.

# - Initial Vocabulary: {"b", "a", "n", " ", "d"}(unique chars)

# - Initial State: b a n a n a   b a n d a n a

# Step-by-Step BPE AlgorithmIteration

# 1: Identify the most frequent pairScanning the text b a n a n a   b a n d a n a, we look for the pair of tokens that appears most often.

# - The pair "a" + "n" appears 4 times (twice in "banana", twice in "bandana").
# - The pair "n" + "a" appears 3 times.
# - Action: Replace "an" with a new token ID, say <256>.
# - New Text: b <256> <256> a   b <256> d <256> a
# -Vocabulary Update: 256: "an"

#Iteration 2: Identify the next most frequent pair
# We look at the new text: b <256> <256> a   b <256> d <256> a.
# The pair "b" + "<256>" (which is "b" + "an") appears 2 times.
# The pair "<256>" + "a" appears 2 times.
# Action: Replace "b<256>" with a new token ID, say <257>.
# New Text: <257> <256> a   <257> d <256> a
# Vocabulary Update: 257: "ban" (Note: This is technically <257>: "b" + <256>)


# Iteration 3: Identify the next most frequent pair

# We look at the new text: <257> <256> a   <257> d <256> a.The pair "<257>" + " " (space) appears 0 times.

# The pair "<256>" + "a" appears 2 times.
#Action: Replace "<256>a" with a new token ID, say <258>.
#New Text: <257> <258>   <257> d <258>
#Vocabulary Update: 258: "ana" (Note: This is <258>: <256> + "a")
#Final Result Summary

## After just three iterations, our original 14-character string has been compressed into 6 tokens:

# Token IDRepresents256"an"257"ban"258"ana"

#Token ID	Represents

#256 |	"an"
#257 |	"ban"
#258 |	"ana"

```python
import json
import collections
```

## We will not develop a simple end to end BPE encoding and test it with simple encoding



```python
class BPETokenizer:
    def __init__(self, vocab_size=300):
        self.vocab_size = vocab_size
        # Start with base 256 ASCII/Byte characters
        self.vocab = {i: bytes([i]) for i in range(256)}
        self.merges = {}

    def _get_stats(self, ids):
      #Initialize a dictionary for counting pairs
        counts = collections.defaultdict(int)
        for pair in zip(ids, ids[1:]):
            counts[pair] += 1
        return counts

    def _merge(self, ids, pair, idx):
        new_ids = []
        i = 0
        while i < len(ids):
            if i < len(ids) - 1 and (ids[i], ids[i+1]) == pair:
                new_ids.append(idx)
                i += 2
            else:
                new_ids.append(ids[i])
                i += 1
        return new_ids

    def train(self, corpus):
        """Builds the vocabulary by iteratively merging frequent pairs."""
        text_bytes = corpus.encode("utf-8")
        ids = list(text_bytes)

        num_merges = self.vocab_size - 256
        for i in range(num_merges):
            stats = self._get_stats(ids)
            if not stats: break

            # Find the most frequent pair
            best_pair = max(stats, key=stats.get)
            new_id = 256 + i

            # Record the merge and update vocab
            self.merges[best_pair] = new_id
            self.vocab[new_id] = self.vocab[best_pair[0]] + self.vocab[best_pair[1]]

            # Update the sequence with the new token
            ids = self._merge(ids, best_pair, new_id)
            print(f"Iteration {i+1}: Merged {best_pair} into {new_id}")

    def encode(self, text):
        """Converts text string into a list of token IDs."""
        ids = list(text.encode("utf-8"))
        # Apply merges in the exact order they were learned during training
        for pair, new_id in self.merges.items():
            ids = self._merge(ids, pair, new_id)
        return ids

    def decode(self, ids):
        """Converts a list of token IDs back into a UTF-8 string."""
        byte_parts = [self.vocab[idx] for idx in ids]
        return b"".join(byte_parts).decode("utf-8", errors="replace")

    def save_tokenizer(self, filename):
        """Saves merges and vocab to a JSON file."""
        # Convert tuple keys to strings for JSON compatibility
        serializable_merges = {f"{k[0]},{k[1]}": v for k, v in self.merges.items()}
        data = {"merges": serializable_merges, "vocab_size": self.vocab_size}
        with open(filename, "w") as f:
            json.dump(data, f)

    @classmethod
    def load_tokenizer(cls, filename):
        """Loads a tokenizer from a file."""
        with open(filename, "r") as f:
            data = json.load(f)
        instance = cls(vocab_size=data["vocab_size"])
        # Restore merges and reconstruct vocab
        for k, v in data["merges"].items():
            pair = tuple(map(int, k.split(",")))
            instance.merges[pair] = v
            instance.vocab[v] = instance.vocab[pair[0]] + instance.vocab[pair[1]]
        return instance
```

```python
corpus = "the cat in the hat bandana banana"
tokenizer = BPETokenizer(vocab_size=270) # 256 base + 14 merges
tokenizer.train(corpus)
```
