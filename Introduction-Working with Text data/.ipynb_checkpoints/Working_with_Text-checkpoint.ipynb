We will be coverting the following aspects in this course

Phase 1: Core Development & Architecture
Data Preparation & Sampling: The foundation begins by gathering and cleaning massive datasets. This includes sampling the data to ensure variety and quality.

--------------------------------------------------

Attention Mechanisms: Implementation of the mathematical "attention" layers that allow the model to understand the relationship between different words in a sequence.

LLM Architecture: Designing the structural framework (e.g., number of layers, hidden dimensions, and parameters) that defines how the model processes information.

---------------------------------------------------------

Phase 2: Training the Foundation
Training Loop: The model enters a repetitive cycle of processing data, calculating errors (loss), and adjusting its internal parameters to improve.

Foundation Model Creation: The result of the initial training is a "Foundation Model"—a general-purpose AI that understands language but hasn't been specialized yet.

-------------------------------------------------------------

Phase 3: Fine-Tuning & Specialization
From the Foundation Model, the process branches into two specific paths:

Path A: Classification

The Foundation Model is paired with a Dataset with class labels.

It is then trained to become a Classifier (e.g., for sentiment analysis or spam detection).

Path B: Instruction & Assistance

The model undergoes Fine-tuning using an Instruction dataset.

It loads pretrained weights to retain its general knowledge.

The final result is a Personal Assistant (like the chat agents we use today).

## We will first download the verdict short story book

```python
import os
import requests

if not os.path.exists("the-verdict.txt"):
    url = (
        "https://raw.githubusercontent.com/rasbt/"
        "LLMs-from-scratch/main/ch02/01_main-chapter-code/"
        "the-verdict.txt"
    )
    file_path = "the-verdict.txt"

    response = requests.get(url, timeout=30)
    response.raise_for_status()
    with open(file_path, "wb") as f:
        f.write(response.content)

```

## Opening the book

```python
with open("the-verdict.txt", "r", encoding="utf-8") as f:
    raw_text = f.read()

print("Total number of character:", len(raw_text))
print(raw_text[:99])
```

## First we will try to build a simple tokenizer

The goal is to tokenize and embed this text for an LLM

Let's develop a simple tokenizer based on some simple sample text that we can then later apply to the text above

The following regular expression will split on whitespaces

```python
import re

text = "I am going to build gpt from scratch."
result = re.split(r'(\s)', text)

print(result)
```

We don't only want to split on whitespaces but also commas and periods, so let's modify the regular expression to do that as well

```python
result = re.split(r'([,.]|\s)', text)

print(result)
```

As we can see, this creates empty strings, let's remove them
